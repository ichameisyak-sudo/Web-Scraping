#**SCRAPE USING HTTP**

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import re

url = 'https://news.detik.com/berita/d-8647946/bpbd-karhutla-di-kalimantan-selatan-capai-12-ribu-hektare'
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36'
}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')
paragraph_num = []
isi_inti_list = []
article_body = soup.find('div', class_='detail__body-text')
iklan_patterns = [
    r'SCROLL TO CONTINUE WITH CONTENT',
    r'ADVERTISEMENT',
    r'Baca juga:.*',
    r'Simak juga Video.*',
    r'\[Gambas:.*\]'
]

if article_body:
    paragraphs = article_body.find_all('p')
    idx = 1
    for p in paragraphs:
        text = p.text.strip()
        for pattern in iklan_patterns:
            text = re.sub(pattern, '', text, flags=re.IGNORECASE).strip()
        text = re.sub(r'^Jakarta\s*-\s*', '', text)
        if text and not p.find('a'):
            paragraph_num.append(f"Paragraf {idx}")
            isi_inti_list.append(text)
            idx += 1

df = pd.DataFrame({
    'Bagian': paragraph_num,
    'Inti Paragraf': isi_inti_list
})

df
```

# **SCRAPE USING API**

```python
url = "https://api.spaceflightnewsapi.net/v4/articles/"
page = requests.get(url)
data = page.json()
```

```python
articles = data["results"]
articles
```

```python
data_list = []

for item in articles:
    data_list.append({
        "title": item["title"],
        "news_site": item["news_site"],
        "published_at": item["published_at"],
        "url": item["url"],
        "summary": item["summary"]
    })

df = pd.DataFrame(data_list)
df
```
